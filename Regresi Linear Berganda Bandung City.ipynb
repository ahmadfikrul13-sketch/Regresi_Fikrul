## **REGRESI LINEAR KOTA BANDUNG**

Variabel Y = KEMISKINAN

Variabel X(1) = bm_pendidikan

Variabel X(2) = bm_kesehatan

```python
# IMPORT LIBRARY
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go
```

```python
#MEMBACA DATA & MEMILIH KOTA
URL = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv"
try:
    data_all = pd.read_csv(URL)
except Exception:
    data_all = pd.read_csv("dataset_fiskal_97_hipoFull.csv")

df = data_all[data_all["Kota"] == "Kota Bandung"].reset_index(drop=True)

# Ambil hanya variabel yang dipakai
df = df[["Tahun", "bm_pendidikan", "bm_kesehatan", "KEMISKINAN"]]
print(df)
print("\nJumlah observasi:", len(df))
```

```python
# VARIABEL INDEPENDEN & DEPENDEN

X = df[["bm_pendidikan", "bm_kesehatan"]]
y = df["KEMISKINAN"]

X_model = sm.add_constant(X)

model = sm.OLS(y, X_model)
results = model.fit()

print("\n")
print("=" * 50)
print("HASIL REGRESI LINEAR KOTA BANDUNG")
print("=" * 50)
print(results.summary())


# MENGAMBIL KOEFISIEN REGRESI
intercept = results.params["const"]
coef_dbh = results.params["bm_pendidikan"]
coef_dak = results.params["bm_kesehatan"]

r_squared = results.rsquared
adjusted_r_squared = results.rsquared_adj

# MEMBUAT PERSAMAAN REGRESI
persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_dbh:.3f}) X₁"
    f" + ({coef_dak:.3f}) X₂"
)

print("\n")
print("=" * 50)
print("                    PERSAMAAN REGRESI")
print("=" * 50)
print(persamaan)
print("X₁ = bm_pendidikan, X₂ = bm_kesehatan, Y = KEMISKINAN")

print(f"\nR²            = {r_squared:.4f}")
print(f"Adjusted R²   = {adjusted_r_squared:.4f}")
print(f"F-statistic   = {results.fvalue:.3f} (p = {results.f_pvalue:.4f})")


# MENGHITUNG NILAI PREDIKSI
df["Prediksi"] = results.predict(X_model)
df["Residual"] = df["KEMISKINAN"] - df["Prediksi"]

print("\n")
print("=" * 70)
print("                  DATA AKTUAL & PREDIKSI")
print("=" * 70)
print(df[["Tahun", "bm_pendidikan", "bm_kesehatan", "KEMISKINAN", "Prediksi", "Residual"]].round(2))


# MEMBUAT GRID UNTUK BIDANG REGRESI
x1 = np.linspace(df["bm_pendidikan"].min(), df["bm_pendidikan"].max(), 40)
x2 = np.linspace(df["bm_kesehatan"].min(), df["bm_kesehatan"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)

# MENGHITUNG NILAI Y PADA BIDANG REGRESI
Y = intercept + coef_dbh * X1 + coef_dak * X2


# MEMBUAT FIGURE 3D
fig = go.Figure()


# MENAMPILKAN DATA AKTUAL
fig.add_trace(
    go.Scatter3d(
        x=df["bm_pendidikan"],
        y=df["bm_kesehatan"],
        z=df["KEMISKINAN"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Tahun {t}" for t in df["Tahun"]],
        hovertemplate=(
            "%{text}"
            "bm_pendidikan: %{x}"
            "bm_kesehatan: %{y}"
            "KEMISKINAN: %{z}"
            ""
        ),
    )
)


# MENAMPILKAN BIDANG REGRESI
fig.add_trace(
    go.Surface(
        x=X1, y=X2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Regression Plane",
        hovertemplate=(
            "bm_pendidikan: %{x:.2f}"
            "bm_kesehatan: %{y:.2f}"
            "Prediksi KEMISKINAN: %{z:.2f}"
            ""
        ),
    )
)
```

```python
#MENGATUR TAMPILAN GRAFIK
fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR KOTA BANDUNG"
            ""
            f"{persamaan} | "
            f"R² = {r_squared:.3f}"
        ),
        "x": 0.5,
    },
    scene={
        "xaxis": {"title": "bm_pendidikan (X₁)"},
        "yaxis": {"title": "bm_kesehatan (X₂)"},
        "zaxis": {"title": "KEMISKINAN (Y)"},
    },
    width=1000,
    height=750,
    margin={"l": 0, "r": 0, "b": 0, "t": 100},
    legend={"x": 0.02, "y": 0.98},
)


# MENAMPILKAN GRAFIK
fig.show()
fig.write_html("regresi_3d_kota_bandung.html")
```
